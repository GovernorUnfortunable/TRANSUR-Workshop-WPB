## Introducción

Este documento describe el flujo de trabajo para WPB_BIBLM (Bibliometría - OpenAlex), módulo de TRANSUR que recupera, analiza e itera sobre datasets bibliométricos con OpenAlex como fuente.

**Alcance:** No requiere  expertise técnica profunda, y permite ejecutar búsquedas en OpenAlex de forma iterativa, perfeccionando las mismas a partir de la supervisión humana.

## Narrativa del Problema (Rule 1: Tell a Story)

### ¿Qué hace WPB_BIBLM?

TRANSUR estudia tres núcleos teóricos en circulación global (1960-2026):
- **Núcleo A:** Decolonialidad-Poscolonialidad
- **Núcleo B:** Cosmologías Indígenas
- **Núcleo C:** Teoría de la Dependencia

El módulo WPB_BIBLIM y su script WPB_BIBLM_S01.py, generan un dataset bibliométrico que captura:
1. **Publicaciones relevantes** en Francia y Alemania (recurriendo a la afiliación de autores)
2. **Evolución temporal** (¿cuándo ganó atención?)
3. **Redes de autores** (¿quiénes conversan sobre el tema?) <!-- guarda con esto, no se si lo hacemos realmente-->
4. **Términos** (¿con qué lenguaje se habla del tema?) <!-- guarda con esto, no se si lo hacemos realmente-->

### Problema: Exploración manual vs. sistemática

Búsqueda manual en OpenAlex, via web, puede ser:
- Lenta (accionar recurrente y repetitivo)
- Irreproducible (\"creo que dejé de buscar... aquí...\")
- Opaca (\"¿cómo había encontrado aquello?\")

**Lógica:** Un mismo script que, parametrizado y automatizado, permite iteración estable, rápida y con trazabilidad del proceso. 


## Arquitectura del Flujo (Rule 2: Document the Process)

### El Ciclo Iterativo

```
INICIO
  ↓
[0] Preparar YAML + Diccionario CSV
  ↓
[1] Ejecutar: python WPB_BIBLM_S01-export.py --config WPB_BIBLM.yaml
  ↓
[2] Generar:
    - Dataset CSV (registros bibliográficos)
    - Estadísticas (básicas + avanzadas)
    - Candidatos (términos nuevos sugeridos)
  ↓
[3] Usuario REVISA:
    - ¿Suficientes registros? (≥500 es inicio razonable)
    - ¿Balanceado temporal/geográfico?
    - ¿Ruido? (disciplinas irrelevantes)
    - ¿Buenos candidatos de términos?
  ↓
[4] DECISIÓN:
    - SI → Iterar: editar CSV candidatos, re-ejecutar
    - NO → Dataset es estable, proceder a análisis
```

### Decisiones Arquitectónicas Clave

| Decisión | Razón | Trade-off |
|----------|-------|----------|
| OpenAlex API | Cobertura global, multiidioma, acceso público | Calidad variable vs. WoS/Scopus |
| Búsqueda + stemming | Captura variantes (decolonial, decolonialidad) | Más ruido; requiere filtrado |
| Extracción de candidatos | Descubrimiento automático de sinónimos | No captura semántica profunda |
| Reintentos 3x exponential backoff | Robusto ante fallos transitorios | No maneja límites permanentes |
| NLTK stopwords | Librería mantenida, multiidioma | Dependencia externa |


## Cómo Usar WPB_BIBLM (Rule 7: Build a Pipeline)

### Prerequisitos

```bash
pip install -r requirements.txt
```

Incluye: `requests`, `pyyaml`, `nltk`

### Paso 1: Preparar archivos

Necesitas:
1. `WPB_BIBLM.yaml` — configuración (parametrizable)
2. `palabras_clave_biblm.csv` — diccionario (editables por usuario)
3. `WPB_BIBLM_S01-export.py` — script (no tocar)

### Paso 2: Ejecutar búsqueda

```bash
python WPB_BIBLM_S01-export.py --config WPB_BIBLM.yaml
```

Genera en `output/`:
- `WPB_BIBLM_00X_v0.Y.csv` — dataset bibliográfico
- `WPB_BIBLM_00X_estadisticas_basicas.txt`
- `WPB_BIBLM_00X_estadisticas_avanzadas.txt`
- `WPB_BIBLM_00X_CANDIDATOS_sintetizado.csv`
- `WPB_BIBLM_00X_CANDIDATOS_temporal.csv`

### Paso 3: Revisar resultados

Abre:
1. `estadisticas_basicas.txt` — ¿suficientes registros?
2. `CANDIDATOS_sintetizado.csv` — ¿buenos términos?
3. `CANDIDATOS_temporal.csv` — ¿patrón de atención?

### Paso 4: Decidir iteración

**SI ITERAR:**
- Abre `WPB_BIBLM_00X_CANDIDATOS_sintetizado.csv`
- Marca columna `seleccionar = 1` para términos a agregar
- Script agrega automáticamente a v0.1
- Re-ejecuta:

```bash
python WPB_BIBLM_S01-export.py --config WPB_BIBLM.yaml
```

**SI NO ITERAR:**
- Dataset es ESTABLE
- Documenta en README.md por qué no iterar
- Procede a siguiente fase (análisis cualitativo)
